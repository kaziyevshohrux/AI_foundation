```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4])
print(arr1)
```

```python
#shape
arr1.shape

arr2 = np.array([[1,23,32], [54, 34,43]])
print(arr2)
print(arr2.shape)
```

```python
#dtype

a = np.array([32,23,4433,4,44434434444344444])
print(a.dtype)
```

```python
# reshape

a = np.array([1,2,3,4,5,6])

c = a.reshape(3,2)
print(c)
```

```python
#astype
a = np.array([1,2,3,4,5,6])

b = a.astype(float)
print(b)
```

```python
# zeros   , ones

z = np.zeros((2,4))
o = np.ones((3,3))

print("zeros:", z)
print("ones:", o)
```

```python
# ndim

a = np.array([[1],[9]])

print("oned:", a.ndim)
```

```python
#arange


print(np.arange(2,99))
```

```python
# sum , mean, min , max

a = np.array([1,2,3,4,5,6])

print(a.sum())
print(a.min())
print(a.max())
print(a.mean())
```

```python
# flatten

a = np.array([[2,2],[3,3],[4,4]])
print(a.flatten())
```

```python
# transpose T

a = np.array([[1,2],[3,4]])

print(a.T)
```

```python
a = np.array([2,2])
b = np.array([2,10])

print(a.dot(b))
```

```python
#random.rand()

a= np.random.rand(2,3)
print(a)
```
