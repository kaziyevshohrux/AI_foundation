```python
data_with_missing = {
    'cars': ["BMW", "Volvo", None, "Tesla", "Ferrari", None, "BMW"],
    'counts': [3, 7, None, 3, None, None, 6]
}

df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
df_missing.isnull().sum() #null bulganlarni boolean qilib chiqaradi
```

```python
df_missing.dropna(how='all') # Null bulgan ma'lumotlarni olib tashlayapdi data cleaning
```

```python
df_missing.dropna(subset='cars')
```

```python
df_missing.fillna({'cars': 'unkown', 'counts': 0}) #missing value larga value qoyib chiqadi
```
