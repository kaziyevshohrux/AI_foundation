```python
print("create new columns preview (math , reading , writing score) ")

df["total score"] = df[["math score", "reading score", "writing score"]].sum(axis=1)  #create total score
df["avarge score"] = df["total score"] / 3
df["passed"] = np.where(df["avarge score"] >=50, "Yes", "No")

print(df[["total score", "avarge score", "passed"]].head())
```

```python
test_prep_group = df.groupby("test preparation course")[["math score", "reading score", "writing score"]].mean()
print("Students avarge score by test preparation:")
print(test_prep_group)


#gender

gender_avg = df.groupby("gender")[["math score", "reading score", "writing score"]].mean()
print("Avarge score by gender")
print(gender_avg)

#passed

passed_rate = df["passed"].value_counts(normalize=True) *100
print("Passed rate")
print(passed_rate)

gender_pass = df.groupby("gender")["passed"].value_counts(normalize=True) * 100
print("Gender passed persenatege")
print(gender_pass)
```

```python
# Concert scores to Numpy array
scores = df[["math score", "reading score", "writing score"]].values

# Mean , std
print("Numpy mean:" , np.mean(scores, axis=0))
print("Numpy std", np.std(scores, axis=0))

# Z-score (normalize)
# avarge = 0
z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)
print("Z-score", z_score[:5])
```

```python
# how many students take best scores

top_scores = df[df["avarge score"] >= 90]
print("Top students score above 90:", len(top_scores))

# how many students are failes

failed_students = df[df["passed"] == "No"]
print("Failed students:" , len(failed_students))
```
